```python
# Importing the necessary files
import pandas as pd
# This code makes charts appear in the notebook in the first run
%matplotlib inline
```

```python
df = pd.read_csv('https://www.dropbox.com/scl/fi/jgm1tvjtin4yv8vw6g0ct/city_populations_2025_updated.csv?rlkey=d71j1pfly9kndmez3b27kqa63&raw=1')
df
```

```python
# Only include years above a certain limit (otherwise takes ages!)
df2 = df[df['Year']>1991]
df2.reset_index()
```

```python
# Pivot data
df3 = df2.pivot(index='Year',
                          columns='City',
                          values=['Population'])
df3 = df3['Population']# If we don't do this, then the word population will appear next to the city in the BCR
df3
```

```python
import bar_chart_race as bcr
bcr.bar_chart_race(df=df3, filename=None,
                   n_bars=12, # the package is not very good at handling too many changes, you will see that
                   perpendicular_bar_func='median',
                   title='Most Populous Cities in the World')
```
